### Moving Shape Horizontally 🎨
Using a while loop, we can make a circle move horizontally across the canvas. By checking the edges of the canvas, the circle will bounce back whenever it hits a wall. This demonstrates basic animation, position updates, and collision detection in a simple, visual way.

```python
# Create canvas
canvas = Canvas(width=400, height=300)
canvas.layout.border = "3px solid white"
display(canvas)

# Circle properties
x = 200; y = 150
radius = 20

dx = 3   # horizontal speed
bounces = 0
max_bounces = 3 # stop after 10 bounces

# Animation loop
while bounces < max_bounces:
    canvas.clear()
    canvas.stroke_rect(0, 0, canvas.width, canvas.height)  # draw border
    
    # Draw circle
    canvas.fill_style = "royalblue"
    canvas.fill_circle(x, y, radius)
    
    # Update position
    x += dx
    
    # Bounce check
    if x + radius >= canvas.width or x - radius <= 0:
        dx = -dx
        bounces += 1  # count bounce
    
    time.sleep(0.01)

print(f"Animation stopped after {bounces} bounces.")
```

### Moving Shape Vertically 🎨
Using a while loop, we can make a circle move vertically across the canvas. By checking the edges of the canvas, the circle will bounce back whenever it hits a wall.

```python
# Create canvas
canvas = Canvas(width=400, height=300)
canvas.layout.border = "3px solid white"
display(canvas)

# Circle properties
x = 200; y = 150
radius = 20

dy = 3 # vertical speed
bounces = 0
max_bounces = 5 # stop after 10 bounces

# Animation loop
while bounces < max_bounces:
    canvas.clear()
    canvas.stroke_rect(0, 0, canvas.width, canvas.height)  # draw border
    
    # Draw circle
    canvas.fill_style = "royalblue"
    canvas.fill_circle(x, y, radius)
    
    # Update position
    y += dy
    # Bounce check
    if y + radius >= canvas.height or  y - radius <=0:
        dy = -dy
        bounces += 1  # count bounce
    
    time.sleep(0.02)

print(f"Animation stopped after {bounces} bounces.")
```

### Finding the Pressed key 🎨
This example detects which key is pressed on the keyboard.

```python
canvas = Canvas(width=400, height=300)
display(canvas)

canvas.stroke_rect(0,0,canvas.width,canvas.height)

# Keyboard event handler
canvas.font = "18px Arial"
canvas.fill_style = "red"
def handle_keyboard_event(key, shift_key, ctrl_key, meta_key):
    canvas.clear()
    canvas.stroke_rect(0, 0, canvas.width, canvas.height)
    canvas.fill_text(key, 20, 100)

canvas.on_key_down(handle_keyboard_event)
canvas.focus()
```

### Moving Rectangular Pad horizontally  🎨
Moving Rectangular Pad horizontally  across the canvas.

```python
# Create canvas
canvas = Canvas(width=400, height=400)
canvas.layout.border = "3px solid white"
display(canvas)

# Paddle properties
pad_state = {
    'x': 170,
    'y': 350,
    'width': 80,
    'height': 20,
    'step': 10
}

def draw_pad():
    canvas.clear()
    canvas.stroke_rect(0, 0, canvas.width, canvas.height) # Draw border
    
    # Draw paddle
    canvas.fill_style = "red"
    canvas.fill_rect(
        pad_state['x'],
        pad_state['y'],
        pad_state['width'],
        pad_state['height']
    )

canvas.font = "18px Arial"
# Keyboard control
def handle_keyboard_event(key, shift_key, ctrl_key, meta_key):

    if key == "ArrowLeft":
        pad_state['x'] -= pad_state['step']

    elif key == "ArrowRight":
        pad_state['x'] += pad_state['step']

    draw_pad() # updating the new pad
    canvas.fill_text(pad_state['x'], 20, 100)

canvas.on_key_down(handle_keyboard_event)
canvas.focus()   # Enable keyboard control

draw_pad() # Drawing initial pad 
```

### Changing Color while Bouncing 🎨
While boucing the Circle on the wall, the color of the circle will change randomly. 

```python
# Create canvas
canvas = Canvas(width=400, height=300)
canvas.layout.border = "3px white"
display(canvas)

# Circle properties
x = 200
y = 150
radius = 20

dx = 3   # horizontal speed

bounces = 0
max_bounces = 4

# Initial color
circle_color = "royalblue"

# List of possible colors
colors = ["red", "green", "blue", "orange", "purple", "pink", "cyan", 'royalblue','white','black']



# Animation loop
while bounces < max_bounces:
    canvas.clear()
    
    # Draw border
    canvas.stroke_rect(0, 0, canvas.width, canvas.height)
    
    # Draw circle    
    canvas.fill_style = circle_color
    canvas.fill_circle(x, y, radius)
    
    # Update position
    x += dx
      
    # Bounce check - change color on bounce
    if x + radius >= canvas.width or x - radius <= 0:
        dx = -dx
        bounces += 1
        r = random.randint(0,255)
        g = random.randint(0,255)
        b = random.randint(0,255)
        circle_color = f"rgb({r}, {g}, {b})"  # change color

    time.sleep(0.03)

print(f"Animation stopped after {bounces} bounces.")
```

### Bouncing Circle and Moving Rectangular Pad horizontally  🎨
While boucing the Circle and Moving Rectangular Pad horizontally  across the canvas using thread. 

```python
canvas = Canvas(width=400, height=400)
canvas.layout.border = "3px solid white"
display(canvas)

# Paddle properties
pad_state = {
    'x': 170,
    'y': 350,
    'width': 80,
    'height': 20,
    'step': 15
}

# Ball properties
ball = {
    'x': 200,
    'y': 100,
    'r': 10,
    'dx': 3,
    'dy': 3
}

bounce_count = 0
max_bounce = 5

# Draw scene
def draw_scene():
    global circle_color 
    
    circle_color = 'blue'
    canvas.clear()   
    canvas.stroke_rect(0, 0, canvas.width, canvas.height)
    
    # Paddle
    canvas.fill_style = circle_color
    canvas.fill_rect(
        pad_state['x'],
        pad_state['y'],
        pad_state['width'],
        pad_state['height']
    )
    
    # Ball
    canvas.fill_style = circle_color
    canvas.fill_circle(ball['x'], ball['y'], ball['r'])
    
    # Bounce counter
    canvas.fill_style = "white"
    canvas.fill_text(f"Bounces: {bounce_count}", 10, 20)

# Ball animation
def move_ball():
    global bounce_count
    
    while True:

        if bounce_count >= max_bounce:
            canvas.fill_style = "yellow"
            canvas.fill_text("Game Over!", 150, 200)
            break

        ball['x'] += ball['dx']
        ball['y'] += ball['dy']

        # Wall collision
        if ball['x'] - ball['r'] <= 0 or ball['x'] + ball['r'] >= canvas.width:
            ball['dx'] *= -1
            bounce_count += 1
            circle_color = 'white'

        if ball['y'] - ball['r'] <= 0 or ball['y'] + ball['r'] >= canvas.height:
            ball['dy'] *= -1
            bounce_count += 1

        draw_scene()
        time.sleep(0.02)

# Keyboard control
def handle_keyboard_event(key, shift_key, ctrl_key, meta_key):

    if key == "ArrowLeft":
        pad_state['x'] -= pad_state['step']

    elif key == "ArrowRight":
        pad_state['x'] += pad_state['step']

    # Keep paddle inside canvas
    pad_state['x'] = max(0, min(canvas.width - pad_state['width'], pad_state['x']))

    draw_scene()

canvas.on_key_down(handle_keyboard_event)
canvas.focus()

draw_scene()

threading.Thread(target=move_ball, daemon=True).start() # move the ball continuously while responding to keyboard input
```

```python
import random
import time
import ipywidgets as widgets
from IPython.display import display, clear_output

score = 0
rounds = 10

buttons = [widgets.Button(description=str(i), layout=widgets.Layout(width='60px')) for i in range(1,10)]
output = widgets.Output()

target = random.randint(1, 9)

def update_buttons():
    global target
    target = random.randint(1, 9)
    with output:
        clear_output()
        print(f"🎯 Click: {target}")

def on_click(b):
    global score, rounds
    if rounds <= 0:
        return
    
    if int(b.description) == target:
        score += 1
        print("✅ Correct!")
    else:
        print("❌ Wrong!")
    
    rounds -= 1
    time.sleep(0.5)
    
    if rounds == 0:
        with output:
            clear_output()
            print(f"🏁 Game Over! Final Score: {score}/10")
    else:
        update_buttons()

for b in buttons:
    b.on_click(on_click)

display(output)
display(widgets.HBox(buttons))

update_buttons()
```

```python
from ipycanvas import Canvas
from IPython.display import display
import time
import threading
import random

canvas = Canvas(width=400, height=400)
canvas.layout.border = "3px solid white"
display(canvas)

# Paddle properties
pad_state = {
    'x': 170,
    'y': 350,
    'width': 80,
    'height': 20,
    'step': 15
}

# Ball properties
ball = {
    'x': 100,
    'y': 100,
    'r': 15,
    'dx': 3,
    'dy': 3,
    'color': "hsl(0, 100%, 50%)"   # initial color
}

bounce_count = 0
max_bounce = 5
clr = 'white'
# Draw scene
def draw_scene():
        
    canvas.clear()
    canvas.stroke_rect(0, 0, canvas.width, canvas.height)
    
    # Paddle
    canvas.fill_style = "red"
    canvas.fill_rect(
        pad_state['x'],
        pad_state['y'],
        pad_state['width'],
        pad_state['height']
    )
    
    # Ball
    canvas.fill_style = ball['color']
    canvas.fill_circle(ball['x'], ball['y'], ball['r'])
    
    # Bounce counter
    #canvas.fill_style = "white"
    canvas.fill_text(f"Bounces: {bounce_count}", 10, 20)


# Ball animation
def move_ball():
    global bounce_count
    global clr
    
    while True:

        if bounce_count >= max_bounce:
            canvas.fill_style = "yellow"
            canvas.fill_text("Game Over!", 150, 200)
            break

        ball['x'] += ball['dx']
        ball['y'] += ball['dy']

        # Wall collision
        if ball['x'] - ball['r'] <= 0 or ball['x'] + ball['r'] >= canvas.width:
            ball['dx'] *= -1
            ball['color'] = f"hsl({random.randint(0,360)}, 40%, 90%)"

        if ball['y'] - ball['r'] <= 0: 
            ball['dy'] *= -1
            ball['color'] = f"hsl({random.randint(0,360)}, 50%, 50%)"
          
        # ball bouncing from paddle 
        if ball['y'] + ball['r']>= pad_state['y'] and pad_state['x']<= ball['x'] <= pad_state['x']+pad_state['width']:
            ball['dy'] *= -1
            ball['color'] = f"hsl({random.randint(0,360)}, 100%, 50%)"
            bounce_count += 1
         
        if ball['y'] >= canvas.height + ball['r']:
            canvas.fill_style = "yellow"
            canvas.fill_text("Lost!", 150, 200)
            break
       
        draw_scene()
        time.sleep(0.03)


# Keyboard control
def handle_keyboard_event(key, shift_key, ctrl_key, meta_key):

    if key == "ArrowLeft":
        pad_state['x'] -= pad_state['step']

    elif key == "ArrowRight":
        pad_state['x'] += pad_state['step']

    # Keep paddle inside canvas
    pad_state['x'] = max(0, min(canvas.width - pad_state['width'], pad_state['x']))

    draw_scene()

canvas.on_key_down(handle_keyboard_event)
canvas.focus()

draw_scene()

threading.Thread(target=move_ball, daemon=True).start()
```
